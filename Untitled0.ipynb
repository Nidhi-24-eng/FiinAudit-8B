````python
from unsloth import FastLanguageModel
import torch

MODEL_NAME     = "unsloth/Meta-Llama-3.1-8B-Instruct"
MAX_SEQ_LENGTH = 2048

print(f"Loading {MODEL_NAME} in 4-bit QLoRA...")
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name     = MODEL_NAME,
    max_seq_length = MAX_SEQ_LENGTH,
    dtype          = None,      # auto: fp16 on T4
    load_in_4bit   = True,      # QLoRA — keeps VRAM < 10 GB
)
print("Base model loaded.")

# Apply QLoRA adapters to all projection layers
model = FastLanguageModel.get_peft_model(
    model,
    r                          = 16,
    lora_alpha                 = 16,
    lora_dropout               = 0.05,
    target_modules             = ["q_proj","k_proj","v_proj","o_proj",
                                   "gate_proj","up_proj","down_proj"],
    bias                       = "none",
    use_gradient_checkpointing = "unsloth",  # saves ~30% VRAM
    random_state               = 42,
)
model.print_trainable_parameters()

used  = torch.cuda.memory_allocated() / 1024**3
total = torch.cuda.get_device_properties(0).total_memory / 1024**3
print(f"\nVRAM used: {used:.1f} GB / {total:.1f} GB  ({used/total*100:.0f}%)")
print("Ready for training!" if used < 10 else "WARNING: High VRAM usage")
````

````python
import json
from datasets import Dataset
from pathlib import Path

DATA_DIR = Path("/content/data")

SYSTEM_PROMPT = (
    "You are FinAudit-8B, a domain-adapted financial analysis model. "
    "Your task is to extract structured financial metrics and risk information "
    "from SEC 10-K filings and financial reports. "
    "ALWAYS respond with a single, valid JSON object matching the required schema. "
    "Return null for any field not explicitly mentioned in the source text. "
    "Do NOT hallucinate numbers. Do NOT add commentary outside the JSON object."
)

def load_jsonl(path):
    with open(path) as f:
        return [json.loads(line) for line in f if line.strip()]

def fmt(sample):
    msgs = [
        {"role": "system",    "content": sample.get("system", SYSTEM_PROMPT)},
        {"role": "user",      "content": sample["instruction"]},
        {"role": "assistant", "content": sample["output"]},
    ]
    return tokenizer.apply_chat_template(
        msgs, tokenize=False, add_generation_prompt=False
    )

train_raw = load_jsonl(DATA_DIR / "train.jsonl")
val_raw   = load_jsonl(DATA_DIR / "val.jsonl")

train_ds = Dataset.from_dict({"text": [fmt(s) for s in train_raw]})
val_ds   = Dataset.from_dict({"text": [fmt(s) for s in val_raw]})

print(f"Train: {len(train_ds):,} samples")
print(f"Val:   {len(val_ds):,} samples")
print(f"\nSample preview (first 300 chars):")
print(train_ds[0]["text"][:300] + "...")
````

````python

from trl import SFTTrainer, SFTConfig
import time, os, shutil
from pathlib import Path

SFT_OUT    = "/content/checkpoints/sft-adapter"
DRIVE_ROOT = Path("/content/drive/MyDrive/FinAudit-8B")
os.makedirs(SFT_OUT, exist_ok=True)

gpu      = torch.cuda.get_device_properties(0).name
use_bf16 = any(x in gpu for x in ("A100","A10","L4","H100"))
use_fp16 = not use_bf16
print(f"GPU: {gpu}  |  fp16={use_fp16}  bf16={use_bf16}")

sft_config = SFTConfig(
    output_dir                  = SFT_OUT,
    num_train_epochs            = 3,
    per_device_train_batch_size = 2,
    gradient_accumulation_steps = 4,
    warmup_ratio                = 0.03,
    learning_rate               = 2e-4,
    weight_decay                = 0.01,
    lr_scheduler_type           = "cosine",
    fp16                        = use_fp16,
    bf16                        = use_bf16,
    logging_steps               = 10,
    eval_strategy               = "steps",
    eval_steps                  = 100,
    save_strategy               = "steps",
    save_steps                  = 200,
    save_total_limit            = 2,
    load_best_model_at_end      = True,
    metric_for_best_model       = "eval_loss",
    report_to                   = "none",
    seed                        = 42,
    dataloader_num_workers      = 2,
    dataset_text_field          = "text",
    max_seq_length              = MAX_SEQ_LENGTH,
    dataset_num_proc            = 2,
    packing                     = True,
)

trainer = SFTTrainer(
    model         = model,
    tokenizer     = tokenizer,
    train_dataset = train_ds,
    eval_dataset  = val_ds,
    args          = sft_config,
)

print(f"\nStarting SFT training...")
print(f"  3 epochs | effective batch=8 | packing=True")
print(f"  Est. time: 90-120 min on T4")
print(f"  Checkpoints auto-saved every 200 steps\n")

t0   = time.time()
stat = trainer.train()
mins = (time.time() - t0) / 60

print(f"\n{'='*50}")
print(f"  SFT COMPLETE in {mins:.1f} min")
print(f"  Final loss: {stat.training_loss:.4f}")
print(f"{'='*50}")

# Auto-save adapter to Drive
model.save_pretrained(SFT_OUT)
tokenizer.save_pretrained(SFT_OUT)
drive_sft = str(DRIVE_ROOT / "checkpoints/sft-adapter")
if os.path.exists(drive_sft): shutil.rmtree(drive_sft)
shutil.copytree(SFT_OUT, drive_sft)
size_mb = sum(os.path.getsize(os.path.join(d,f))
              for d,_,fs in os.walk(SFT_OUT) for f in fs) / 1024**2
print(f"Adapter saved to Drive ({size_mb:.0f} MB)")
````

````python
FastLanguageModel.for_inference(model)
import time, json

TEST = (
    "Apple Inc. reported net revenue of $394.3 billion for fiscal year 2023. "
    "Net income was $97.0 billion, diluted EPS $6.13. "
    "Key risks: supply chain concentration, FX headwinds, smartphone competition."
)
msgs = [
    {"role": "system", "content": SYSTEM_PROMPT},
    {"role": "user",   "content": f"Extract financial metrics as JSON:\n{TEST}"},
]
prompt = tokenizer.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)
inp    = tokenizer(prompt, return_tensors="pt").to("cuda")

t0 = time.time()
with torch.no_grad():
    out = model.generate(**inp, max_new_tokens=400, temperature=0.1,
                         do_sample=True, pad_token_id=tokenizer.eos_token_id)
ms = (time.time() - t0) * 1000

resp = tokenizer.decode(out[0][inp["input_ids"].shape[1]:], skip_special_tokens=True)
print(f"Latency: {ms:.0f} ms\n")
print(resp)

try:
    parsed = json.loads(resp)
    print("\nValid JSON!")
    for k, v in parsed.items():
        if v is not None: print(f"  {k}: {v}")
except json.JSONDecodeError:
    print("\nNot valid JSON yet — ORPO will fix this!")
````

````python
import random
from datasets import Dataset

random.seed(42)

PROSE_REJECTED = [
    "Based on my analysis the company appears financially healthy. I recommend further due diligence.",
    "The filing discusses several metrics. Revenue looks healthy and management expressed cautious optimism.",
    "I cannot determine exact figures from this excerpt. The company faces typical regulatory risks.",
]

def make_pairs(samples, n=2000):
    pairs = []
    random.shuffle(samples)
    for s in samples[:n]:
        prompt = (
            f"<|system|>\n{s.get('system', SYSTEM_PROMPT)}\n"
            f"<|user|>\n{s['instruction']}\n<|assistant|>\n"
        )
        chosen = s["output"]
        r = random.random()

        if r < 0.45:
            rejected = random.choice(PROSE_REJECTED)
        elif r < 0.70:
            try:
                obj = json.loads(chosen)
                for k, v in obj.items():
                    if isinstance(v, (int, float)) and v is not None:
                        obj[k] = random.choice([999.9, 0.01, -42.5, 12345.0])
                rejected = json.dumps(obj, indent=2)
            except Exception:
                rejected = '{"revenue_usd_millions": 999.9}'
        elif r < 0.85:
            try:
                obj  = json.loads(chosen)
                half = {k: obj[k] for k in list(obj)[:len(obj)//2]}
                rejected = json.dumps(half, indent=2)[:-5]
            except Exception:
                rejected = '{"error": "extraction failed"'
        else:
            rejected = f"```json\n{chosen}\n```\nHope this helps!"

        pairs.append({"prompt": prompt, "chosen": chosen, "rejected": rejected})
    return pairs

train_raw = load_jsonl(DATA_DIR / "train.jsonl")
pairs     = make_pairs(train_raw, n=2000)
orpo_ds   = Dataset.from_list(pairs)

prose_n = sum(1 for p in pairs if not p["rejected"].strip().startswith("{"))
print(f"Preference pairs: {len(orpo_ds):,}")
print(f"  Prose rejections: {prose_n}")
print(f"  JSON rejections:  {len(pairs)-prose_n}")
print(f"\nChosen  (80): {pairs[0]['chosen'][:80]}...")
print(f"Rejected (80): {pairs[0]['rejected'][:80]}...")
````
